```python
# Optional: use the T4 for Replicator-style SO(3) (cos θ = 1/3)
try:
    import torch
    if torch.cuda.is_available():
        print(f"[+] T4 GPU: {torch.cuda.get_device_name(0)}")
        theta = torch.acos(torch.tensor(1/3.0)).item()  # cos θ = 1/3
        # Rotation about x by θ (Replicator α)
        c, s = torch.cos(torch.tensor(theta)), torch.sin(torch.tensor(theta))
        Rx = torch.tensor([[1, 0, 0], [0, c.item(), -s.item()], [0, s.item(), c.item()]], device='cuda', dtype=torch.float32)
        v = torch.tensor([1.0, 0.0, 0.0], device='cuda')
        out = Rx @ v
        print(f"[+] Replicator α (x-rotation θ=arccos(1/3)) on GPU: {out.tolist()}")
    else:
        print("No GPU. Enable T4: Runtime → Change runtime type → T4 GPU.")
except Exception as e:
    print(f"PyTorch not installed or no GPU: {e}")
    print("Replicator (above) runs on CPU. Enable T4 in Colab for this cell.")
```
